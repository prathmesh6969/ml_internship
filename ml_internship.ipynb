```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Step 1: Load the dataset
df = pd.read_csv("city_day.csv")
df.head()
```

```python
df['City'].unique()
```

```python
plt.figure(figsize=(50,30))
sns.barplot(x='City',y='PM2.5',data=df)
plt.title('City vs PM2.5')
plt.xlabel('City')
plt.ylabel('PM2.5')
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(20,10))
sns.scatterplot(x='NO2',y='PM2.5',data=df,hue='City')
plt.title('NO2 vs PM2.5')
plt.xlabel('NO2')
plt.ylabel('PM2.5')

plt.show()
```

```python
x = [df['PM2.5'].mean(), df['PM10'].mean(), df['CO'].mean(), df['SO2'].mean()]
labels = ['PM2.5','PM10','CO','SO2']

fig, ax = plt.subplots()
ax.pie(x, labels=labels)
ax.set_title('Average of PM2.5, PM10, CO, SO2')
plt.show()
plt.tight_layout()
```
